```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("../predicts/nn_balance5.csv", index_col=0)
df2 = pd.read_csv("../predicts/nn_balance13.csv", index_col=0)
df3 = pd.read_csv("../predicts/nn_balance14.csv", index_col=0)
```

```python
final = []
for i in range(0, df1.shape[0]):
    tmp = [df1['casualty_severity'].iloc[i], df2['casualty_severity'].iloc[i], df3['casualty_severity'].iloc[i]]
    pred = np.median(tmp)
    final.append(int(pred))

print(final)
```

```python
X_testFinal = pd.read_csv('../data/X_test.csv', index_col=0)
my_file = open("../predicts/bagging_nn1.csv", "w+")

my_file.write("Id,casualty_severity\n")
my_i = 0
for i, row in X_testFinal.iterrows():
    my_file.write(f"{i}, {final[my_i]}\n")
    my_i = my_i+1
my_file.close()
```
